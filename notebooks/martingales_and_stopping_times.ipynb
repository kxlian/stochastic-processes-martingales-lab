# Martingales and Optional Stopping

This notebook illustrates a bounded stopping time for a symmetric random walk.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path('..').resolve() / 'src'))

from stochastic_lab.random_walk import symmetric_random_walk
from stochastic_lab.stopping_times import bounded_first_exit_time, terminal_at_stopping_time
from stochastic_lab.martingales import optional_stopping_summary
from stochastic_lab.plotting import plot_optional_stopping
```

```python
walks = symmetric_random_walk(n_steps=500, n_paths=2000, seed=7)
tau = bounded_first_exit_time(walks, boundary=12, max_time=500)
summary = optional_stopping_summary(walks, tau)
summary
```

```python
stopped_values = terminal_at_stopping_time(walks, tau)
plot_optional_stopping(stopped_values, '../figures/optional_stopping_notebook.png')
```

With a bounded stopping time, the empirical stopped mean remains close to the initial value. This is a numerical illustration of optional stopping, not a proof.
