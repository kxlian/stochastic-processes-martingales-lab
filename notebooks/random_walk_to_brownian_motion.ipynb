# Random Walks to Brownian Motion

This notebook visualizes diffusive scaling: S_[nt] / sqrt(n).

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path('..').resolve() / 'src'))

from stochastic_lab.random_walk import symmetric_random_walk, scaled_random_walk
from stochastic_lab.plotting import plot_random_walk_scaling
```

```python
paths = symmetric_random_walk(n_steps=1000, n_paths=20, seed=42)
scaled = scaled_random_walk(paths)
scaled.head()
```

```python
plot_random_walk_scaling(scaled, '../figures/random_walk_scaling_notebook.png', max_paths=10)
```

The visual point is not that a finite random walk is Brownian motion. The point is that the diffusive scaling creates the correct variance scale for convergence intuition.
